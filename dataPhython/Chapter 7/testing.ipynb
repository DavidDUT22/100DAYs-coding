```python
%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt

# Step 5 (part 1): set the default figure size BEFORE creating the figure
plt.rc('figure', figsize=(10, 8))

# Step 1: 1x3 grid of subplots
fig, axes = plt.subplots(1, 3)

axes[0].plot(np.arange(0, 5), 'g--')    # Step 3: green dashed line
axes[1].plot(np.arange(0, 10, 2))
axes[2].plot(np.random.randn(5))

# Step 2: spacing between subplots
fig.subplots_adjust(wspace=0.3, hspace=0.3)

# Step 3: title and axis labels for subplot 1
axes[0].set_title('Subplot 1')
axes[0].set_xlabel('X-axis')
axes[0].set_ylabel('Y-axis')

# Step 4: annotation with arrow and title for subplot 2
axes[1].annotate('Peak', xy=(4, 6), xytext=(2, 7.5),
                 arrowprops=dict(facecolor='black', shrink=0.05))
axes[1].set_title('Subplot 2')

# Step 5 (part 2): save the figure (before show)
fig.savefig('custom_plot.png')

# Step 6: display
plt.show()
```

```python
%matplotlib inline
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Step 1: DataFrame of cumulative sums, indexed 0, 10, 20, ... 90
df = pd.DataFrame(
    np.random.rand(10, 3).cumsum(axis=0),
    columns=['A', 'B', 'C'],
    index=np.arange(0, 100, 10)
)
print(df)

# Step 2: line plot of all columns
ax = df.plot(title='Cumulative Trends', legend=True)
plt.show()

# Step 3: stacked bar plot, figure size (8, 6)
df.plot(kind='bar', stacked=True, figsize=(8, 6), title='Stacked Bar Plot')
plt.show()

# Step 4: normalize each row to sum to 1, then stacked bar plot again
df_norm = df.div(df.sum(axis=1), axis=0)
ax = df_norm.plot(kind='bar', stacked=True, figsize=(8, 6),
                  title='Normalized Stacked Bar Plot')
ax.legend(['A', 'B', 'C'])
plt.show()

# Step 5: Seaborn histogram of column A with 5 bins and KDE
fig, ax = plt.subplots()
sns.histplot(df['A'], bins=5, kde=True, ax=ax, label='Distribution of A')
ax.set_title('Distribution of A')
ax.legend()

# Step 6: save BEFORE show, then display
fig.savefig('hist_a.png')
plt.show()
```
