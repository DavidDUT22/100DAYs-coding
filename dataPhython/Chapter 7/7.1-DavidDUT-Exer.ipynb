```python
## Exercise 7_1: Creating and Customizing Plots with Matplotlib
#1.Create a figure with three subplots arranged in a 1x3 grid using plt.subplots().
# - Plot a simple line for each subplot: subplot 1 with np.arange(0, 5)
# - subplot 2 with np.arange(0, 10, 2)
# - subplot 3 with np.random.randn(5).
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline

plt.rc('figure', figsize=(10, 10))
plt.show

#create 1×3 grid of subplot
fig, x = plt.subplots(1, 3)
x[0].plot(np.arange(0,5), 'g--')
x[1].plot(np.arange(0,10,2), 'r')
x[2].plot(np.random.randn(5), 'b')

#2. adjust spacing between subplots
fig.subplots_adjust(wspace =0.3,hspace =0.3)

#3, title and axis albels for subplot 1
x[0].set_title('Subplot 1')
x[0].set_xlabel('X-axis')
x[0].set_ylabel('Y-axis')

#4, annotation with arrow and tile for subplot2
x[1].annotate('Peak',xy=(4,6), xytext=(2,7),
              arrowprops=dict(facecolor = 'black')) 
x[1].set_title('Subplot 2')

#save file
fig.savefig('JOB-hx_plot.png')

plt.show
```
