```python
# YOUR SOLUTION HERE
def nextStep():
    print("-" * 50)
#1.
print("1. Total revenue and average revenue")
total_revenue = sales['revenue'].sum()
avg_revenue = sales['revenue'].mean()
print(f"Total Revenue:   ${total_revenue}")
print(f"Average Revenue: ${avg_revenue:.2f}")
nextStep()

#2.
print("2. Shop with maximun revenue")
top_shop = sales['revenue'].idxmax()
print(f"The shop with maximun revenue is: {top_shop}")
nextStep()

#3.
print("3. Calculate mean rating using 'skipna=true' and 'false'")
mean_rating_false = sales['rating'].mean(skipna=False)
mean_rating_true = sales['rating'].mean(skipna=True)
print(f"Mean rating using 'skipna=true'  => {mean_rating_true:.2}")
print(f"Mean rating using 'skipna=false' => {mean_rating_false}")
nextStep()

#4. 
print("4. Show 'describe()' for numeric columns")
display(sales.describe())
nextStep()

#5. 
print("5. Compute cumulative revenue in original row order")
cumulative_revenue = sales['revenue'].cumsum()
print(cumulative_revenue)
nextStep()

#6.
print("6. Explain why culumative value depends on row order")
print("A cumulative sum is a running total calculated from top to bottom. Because each row's value")
print("is the sum of that row plus all rows before it, changing the row order changes which values")
print("are added first, resulting in different numbers for each shop.")
```
