```python
#Series re-index
import pandas as pd
import numpy  as np

obj = pd.Series([10, 20, 30], index=['a', 'b', 'c'])
print("Data Before Re-index : ")
print(obj, "\n")
```

```python

obj2 = obj.reindex(['c', 'a', 'b', 'd'])
print("Data After Re-index : ")
print(obj2)
```

```python
#Dataframe re-index
frame = pd.DataFrame({
    'city': ['Phnom Penh', 'Siem Reap', 'Battambang'],
    'population': [2.2, 1.0, 0.3]
}, index=['pp', 'sr', 'bb'])

print("Data Before Re-index : ")
print(frame, "\n")
```

```python

frame2 = frame.reindex(['sr', 'bb', 'pp'])
print("Data Before Re-index : ")
print(frame2, "\n")
```

```python
#Dropping Entries from an Axis
#Remove specific rows or columns.
data = pd.DataFrame(np.arange(16).reshape(4,4),
                    index=['a','b','c','d'],
                    columns=['one','two','three','four'])

print("Data before dropping: ")
print(data, "\n")

```

```python

#drop row
print("Data after dropping: ")
data.drop('c')     # Removes row 'c'

#drop column
data.drop('two', axis=1)
```

### Indexing, Selection, and Filtering
### You can select data by labels or positions, just like arrays, but with more power.

```python
#Example with Series:
obj = pd.Series([3.2, 7.5, -1.5, 3.6], index=['a','b','c','d'])
print(obj, "\n")
print(obj['b'], "\n")       # select by label
print(obj[['a','c']], "\n") # multiple labels
```

```python
print(obj, "\n")

print(obj[1:3], "\n")       # select by position
```

```python
#Label-based slicing includes the end index!
obj['a':'c']
```

```python
#Boolean Indexing
obj[obj > 0]
```

```python
#with dataframes:
print("before boolean: ")
print(data, "\n")

print("after boolean: ")
data[data['one'] > 5]
```

# loc[] and iloc[]
## Two main tools for selecting subsets:
### loc[] → label-based selection
### iloc[] → integer position-based selection

```python
print(data, "\n")
```

```python
loc_data = data.loc['a', ['one','two']]
print(loc_data, "\n")

iloc_data = data.iloc[1, 2]
print(iloc_data)
```

```python
#Arithmetic and Data Alignment
#When performing operations (like +, -, *, /) 
#between Series or DataFrames, pandas automatically aligns data by index labels.
s1 = pd.Series([3, 4, 5], index=['a','b','c'])
print(s1, "\n")

s2 = pd.Series([1, 2, 3], index=['b','c','d'])
print(s2, "\n")
print(s1 + s2)
```

```python
#Arithmetic with Fill Values
#→ replaces missing values with 0 before addition.
s1.add(s2, fill_value=0)
```

```python
#Operations Between DataFrame and Series
#Similar to NumPy broadcasting: the operation applies row by row.
```

```python
frame = pd.DataFrame(np.arange(12).reshape(4,3),
                     columns=list('bcd'),
                     index=['Utah','Ohio','Texas','Oregon'])
print(frame, "\n")
```

```python
series = frame.iloc[0]
print(series, "\n")

print(frame - series)
```

```python
#Function Application and Mapping
#We can apply functions element-wise, row-wise, or column-wise.

frame = pd.DataFrame(np.random.randn(4,3), columns=list('bcd'))
print(frame, "\n")

ufunc = np.abs(frame)        # apply numpy ufunc
print("Frame Data: ")
print(ufunc, "\n")


# apply() across rows or columns
f = lambda x: x.max() - x.min()
print(frame.apply(f), "\n")
print(frame.apply(f, axis='columns'))
```

```python
#applymap() and map():
##applymap() → element-wise for DataFrame
##map() → element-wise for Series
format = lambda x: f'{x:.2f}'
frame.applymap(format)
```

```python
#Sorting and Ranking
#Sorting by index:
frame.sort_index(axis=1, ascending=False)
```

```python
#Sorting by values:
frame.sort_values(by='b')
```

```python
#Ranking:
obj = pd.Series([7, -5, 7, 4, 2, 0, 4])
obj.rank()
```

```python
#Axis Indexes with Duplicate Labels
##Pandas allows duplicate labels but warns that some operations (like reindex) might not work as expected.
obj = pd.Series([1,2,3,4], index=['a','a','b','b'])
print(obj, "\n")

print(obj.index.is_unique)  # False
print(obj['a'])  # returns multiple values
```

```python
#Basic Descriptive Methods
##Pandas includes many summary methods:
df = pd.DataFrame([[1.4, np.nan],
                   [7.1, -4.5],
                   [np.nan, np.nan],
                   [0.75, -1.3]],
                   index=['a','b','c','d'],
                   columns=['one','two'])
print(df, "\n")
print(df.sum(), "\n")              # column-wise sum
print(df.sum(axis=1))        # row-wise sum
```

```python
print(df)

#idxmin() and idxmax()
##Find the index of the minimum or maximum value.
df.idxmax()
```

```python
#describe()
##Generates summary statistics automatically
df.describe()
```
