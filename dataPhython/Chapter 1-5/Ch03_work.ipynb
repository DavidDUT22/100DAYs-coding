### Problem 1: Tuples and Lists
1. Create a tuple person with elements: name (string), age (int), and hobbies (list of 3 strings).
2. Try to modify the age in the tuple (explain the error).
3. Convert the tuple to a list, modify the age, and convert back to a tuple.
4. Use unpacking to assign variables and print the hobbies list.
5. Bonus: Use count() to count occurrences if a hobby is repeated.

```python
#Task : 1  
#person
person = "David", 19, ["Hangout", "Gaming", "Camping", "Sleeping", "Gaming", "Sleeping", "Sleeping"] 
print(person)
```

```python
#Task : 2
#modify age 18->56
person[2]=56
print(person[2])
```

```python
#Task : 3 : 
# convert tuple into list
list_person = list(person)
list_person[1]=56
print(list_person)

# then convert back into tuple
person=list_person
print(person)

```

```python
#task 4
#Use unpacking to assign variables
name = person[0]
age = person[1]
hobbies = person[2]
#print the hobbies list
print(name)
print(age)
print(hobbies)
```

```python
#task 5:
#Bonus: 
sleeping_count = hobbies.count("Sleeping")
hangout_count = hobbies.count("Hangout") 
gaming_count = hobbies.count("Gaming")
camping_count = hobbies.count("Camping") 


print(f"sleeping count is:{sleeping_count}\ngaming count is :{gaming_count}\nhangout count is:{hangout_count}\ncamping count is:{camping_count}")
```
