## A fictional Cambodian online shop dataset
Every group acts as a data analyst helping a shop investigate sales.

```python
sales = pd.DataFrame({
    'shop':['Phnom Penh','Siem Reap','Battambang','Kampot','Takeo','Kandal'],
    'category':['Bag','Watch','Bag','Watch','Bag','Watch'],
    'units':[15,9,20,12,8,18],
    'price_usd':[18,25,20,22,15,24],
    'rating':[4.7,4.2,np.nan,4.6,3.9,4.8]
}).set_index('shop')
sales['revenue'] = sales['units'] * sales['price_usd']
display(sales)
```

```python
print(sales.reindex(['Kampot','Phnom Penh','Pursat']))
print(sales['units'].reindex(['Kampot','Pursat'], fill_value=0))
```

### Student challenge
**Challenge**

```python
# YOUR SOLUTION HERE
#1
n_sales = sales.reindex(['Takeo','Kandal','Phnom Penh','Pursat'])
print(n_sales)
#2
#Pursat has NaN because that label doesn’t exist in sales, so pandas creates the row with no data.
#3
n_units = sales['units'].reindex(['Takeo','Kandal','Phnom Penh','Pursat'], fill_value=0)
print(n_units)
#4
n2_units = sales.reindex(columns=['units','revenue','discount'],fill_value=0)
print(n2_units)
#5
# Sorting only reorders existing rows. Reindexing conforms to a new label set, so it can add, drop and reorder.
```
