```python
# Problem 2: Creating and Manipulating ndarrays
#1. Import NumPy as np. Create a 2D array matrix of shape (3,4) with integers 1-12 using arange and reshape.
#2. Print its shape, ndim, and dtype. Cast to float32.
#3. Perform arithmetic: Add 5 (broadcasting), multiply by a 1D array [1,2,3,4].
#4. Use zeros, ones, and eye to create arrays and explain their uses in signal processing (e.g., identity for transformations).
```

```python
#step 1
#1. Import NumPy as np. Create a 2D array matrix of shape (3,4) with integers 1-12 using arange and reshape.
# create 1D array then convert into 2D array
import numpy as np
array =np.arange(1, 13)
print(array)

matrix = array.reshape(3,4)
print(matrix)
```

```python
#2. Print its shape, ndim, and dtype. Cast to float32.
print(matrix.shape)
print(matrix.ndim)
print(matrix.dtype)

matrix = matrix.astype(np.float32)
print(matrix.dtype)

```

```python
#3. Perform arithmetic: Add 5 (broadcasting), multiply by a 1D array [1,2,3,4].
matrix = matrix +5
print(matrix)
arr = np.arange(1,5)

matrix = matrix * arr
print(matrix)
```

```python
#4. Use zeros, ones, and eye to create arrays and explain their uses in signal processing 
# (e.g., identity for transformations).

zeros = np.zeros((3,4))
print(zeros)
print("\n")

ones =np.ones((3,4))
print(ones)
print("\n")

identity_matrix=np.eye(3)
print(identity_matrix)

```

```python
# Problem 3: Indexing, Slicing, and Boolean Operations
#1. From matrix above, slice the first two rows and last two columns.
print(matrix)
result = matrix[:2, -2:]
print(result)

```

```python
#2. Use boolean indexing to select elements > 5.

result = result >5
print(result)
```

```python
#3. Fancy indexing: Select rows [0,2] and columns [1,3].
print(matrix)

result = matrix[[0,2],[1,3]]
print(result)
```

```python
#4. Transpose the array and compute dot product with itself.
result = matrix.T
print(result)
print('\n')

dotProduct = np.dot(matrix,matrix.T)
print(dotProduct)
```
