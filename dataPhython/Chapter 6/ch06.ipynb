## Handling Missing Values (Dataset ex5)

```python
df = pd.read_csv('ex5.csv')
df
```

```python
df = pd.read_csv('ex5.csv', na_values={'column_name': ['value1', 'value2']})
df
```

## Reading Large Files in Pieces (Dataset ex6)

```python
df = pd.read_csv('ex6.csv', nrows=5)  # Read only 5 rows
chunker = pd.read_csv('ex6.csv', chunksize=1000)  # Read in chunks of 1000 rows
for chunk in chunker:
    # Process each chunk (e.g., count values)
    print(chunk['key'].value_counts())
```
