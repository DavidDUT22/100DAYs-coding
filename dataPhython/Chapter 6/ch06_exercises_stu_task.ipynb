```python
#Exercise 1: task 1
# 1. Create a sample CSV file named "ex1.csv" with the following content (without a header):
# a,b,c,d,message
# 1,2,3,4,hello
# 5,6,7,8,world
# 9,10,11,12,foo
import pandas as pd
import numpy as np

!mkdir Week4_Data
data = [
   ['a','b','c','d','message'],
   [1,2,3,4,'hello'],
   [5,6,7,8,'world'],
   [9,10,11,12,'foo']
]
# convert list into pandas dataFrame
df = pd.DataFrame(data)

#save dataFrame to CSV file without header
df.to_csv('Week4_Data/ex1.csv', header =False, index =False)
```

```python
### 2. Read it into a Pandas DataFrame using read_csv and allow Pandas to assign default column names.
#Read the same "ex1.csv" file again, but this time specify custom column names: ['col1', 'col2', 'col3', 'col4', 'msg'].

df = pd.read_csv('Week4_Data/ex1.csv', header=None)  #use default columns name
df
df =pd.read_csv('Week4_Data/ex1.csv', names=['col1','col2', 'col3', 'col4', 'msg'])
df
```

```python
# Modify the DataFrame by adding a new row with values [13,14,15,16,'bar'] and handle any missing values if they appear (assume none for now).
# Add a new row
df.loc[len(df)] = [13,14,15,16,'bar']
df
# Write the modified DataFrame back to a new CSV file named "out1.csv" without the index, using a semicolon (;) as the delimiter.
df.to_csv('out.csv', index=None, sep=';')
```

```python
#  Read "out1.csv" back into a new DataFrame and print the first two rows to verify.
df= pd.read_csv('out.csv', sep=";")
df
print(df[:2])
```

```python
# 6. Introduce missing values: Replace 'hello' with 'NA' in the DataFrame, 
# then re-write to "out2.csv" and read it back, specifying 'NA' as a custom NA sentinel.
df =df.replace("hello", 'NA')
df
df.to_csv("out2.csv", index=None, sep=";")
df=pd.read_csv("out2.csv", sep=";")
df
```

## Exercise 2: Handling Large Files and Chunks
This exercise covers reading large text files in pieces, iterating over chunks, and aggregating data, based on section 6.1.1.

1. Create a large sample CSV file named "ex6.csv" with 1000 rows of random data, including a 'key' column with values like 'A', 'B', or 'C' (use a loop to generate).

2. Set Pandas display options to show a compact view (max rows=10).

3. Read only the first 5 rows of "ex6.csv" using nrows.

4. Read the file in chunks of 100 rows each using chunksize, and print the shape of the first chunk.

5. Iterate over all chunks and count the occurrences of each unique value in the 'key' column, aggregating into a Series.

6. Write the aggregated counts to a new CSV file named "key_counts.csv".

![alt text](image-1.png)

```python
#Exercise 2:
#1, Create a large sample CSV file named "ex6.csv" with 1000 rows of random data,
#  including a 'key' column with values like 'A', 'B', or 'C' (use a loop to generate).
import random
data = []

for i in range(1000):
    row = {
        'key': random.choice(['A', 'B', 'C']),
        'value': random.randint(1, 10)
    }
    data.append(row)

df = pd.DataFrame(data)
df.to_csv('ex6.csv', index=False)
df

```

```python

# 2. Set Pandas display options to show a compact view (max rows=10).
pd.set_option('display.max_rows', 10)
```

```python
#3, read only 5 first rows
df= pd.read_csv('ex6.csv', nrows=5)
df
```

```python
#4, Read the file in chunks of 100 rows each using chunksize
# and print the shape of the first chunk.
chunker = pd.read_csv('ex6.csv', chunksize=100)
print(pd.read_csv('ex6.csv', nrows =5))
#print first chnk
chunk = next(chunker)
print(chunk)
#print the first chunk shape
print("here is the First chunk shape:")
print(chunk.shape)

```

```python
# 5.Iterate over all chunks and count the occurrences of each unique value in the 'key' column,
#  aggregating into a Series.
# for chunk in chunker:
#     # Process each chunk (e.g., count values)
# #     print(chunk['key'].value_counts())
# # chunker = pd.read_csv('ex.csv', chunksize=100)
for chunk in chunker:
    print(chunk['key'].value_counts())
#task 6  
chunk.to_csv('key_count.csv')

```
